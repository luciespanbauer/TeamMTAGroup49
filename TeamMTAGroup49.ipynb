# Problem 3: Many-Time Pad

In this problem, ten predefined plaintext messages are encrypted using
the same randomly generated key.

This violates the one-time-use requirement of a true One-Time Pad.

The plaintexts, reused key, and ciphertexts are stored in hexadecimal
format.

```python
import secrets
import json

messages = [
    "The meeting starts at noon.",
    "Please bring the project report.",
    "Network security is important.",
    "Never reuse a one-time pad key.",
    "The server is running normally.",
    "Check the logs before changing anything.",
    "Encryption protects confidential data.",
    "A strong key must be random.",
    "Cryptography requires careful implementation.",
    "Key reuse can reveal information."
]

plaintexts = [
    message.encode("ascii")
    for message in messages
]

key_length = max(
    len(message)
    for message in plaintexts
)

reused_key = secrets.token_bytes(key_length)

many_time_ciphertexts = [
    xor_bytes(plaintext, reused_key)
    for plaintext in plaintexts
]

for number, (message, plaintext, ciphertext) in enumerate(
    zip(messages, plaintexts, many_time_ciphertexts),
    start=1
):
    print(f"Message {number}: {message}")
    print(f"Plaintext (hex):  {plaintext.hex()}")
    print(f"Ciphertext (hex): {ciphertext.hex()}")
    print()

print("Reused key (hex):")
print(reused_key.hex())
```

The goal is to exploit the key reuse and recover the plaintext of the
target ciphertext.

The attack consists of:

1. Loading the ciphertexts
2. Defining XOR functions
3. Performing pairwise XOR analysis
4. Identifying likely spaces
5. Recovering initial key bytes
6. Performing crib dragging
7. Recovering additional key bytes
8. Decrypting the target ciphertext

### XOR Functions

The XOR operation is the basis of the attack.

For a many-time pad:

C = M ⊕ K

If two ciphertexts use the same key:

C1 ⊕ C2 = (M1 ⊕ K) ⊕ (M2 ⊕ K)

Since K ⊕ K = 0:

C1 ⊕ C2 = M1 ⊕ M2

```python
def xor_bytes(a, b):
    return bytes(x ^ y for x, y in zip(a, b))

def xor_with_key(ciphertext, key):
    return bytes(c ^ k for c, k in zip(ciphertext, key))
```

### Pairwise XOR Analysis


Each ciphertext is XORed with the other ciphertexts.

Because the same key was reused, the key cancels out:

C1 ⊕ C2 = M1 ⊕ M2

This allows relationships between the plaintexts to be examined without
knowing the key.

```python
pairwise_xors = {}

for i in range(len(ciphertexts)):
    for j in range(i + 1, len(ciphertexts)):

        pairwise_xors[(i + 1, j + 1)] = xor_bytes(
            ciphertexts[i],
            ciphertexts[j]
        )

print("Number of pairwise XORs:", len(pairwise_xors))

for pair, result in list(pairwise_xors.items())[:5]:
    print(
        f"Ciphertexts {pair}: "
        f"{result.hex()}"
    )
```

### ASCII / Space Analysis

The plaintexts are English and contain spaces.

The ASCII value of a space is 0x20.

If a plaintext character is a space:

K = C ⊕ 0x20

Therefore, each ciphertext position can be tested as a possible space.
If the resulting key byte causes the same position in many other
ciphertexts to produce alphabetic characters, this provides evidence
that the original plaintext character was a space.

```python
def is_letter(byte_value):
    return (
        65 <= byte_value <= 90
        or
        97 <= byte_value <= 122
    )


space_candidates = []

for ciphertext_number, ciphertext in enumerate(
    ciphertexts,
    start=1
):

    for position in range(len(ciphertext)):

        possible_key = ciphertext[position] ^ 0x20

        score = 0

        for other_number, other_ciphertext in enumerate(
            ciphertexts,
            start=1
        ):

            if other_number == ciphertext_number:
                continue

            if position < len(other_ciphertext):

                possible_plaintext = (
                    other_ciphertext[position]
                    ^ possible_key
                )

                if is_letter(possible_plaintext):
                    score += 1

        space_candidates.append(
            (
                score,
                ciphertext_number,
                position
            )
        )

space_candidates.sort(reverse=True)

print("Top space candidates:\n")

for score, ciphertext_number, position in space_candidates[:20]:

    print(
        f"Ciphertext {ciphertext_number}, "
        f"position {position}, "
        f"score {score}"
    )
```

## Inital Key Recovery

The highest-scoring space positions from the previous analysis are used
to recover initial key bytes.

For each likely space:

K = C ⊕ 0x20

Unknown key positions are represented by `None`.

```python
space_hypotheses = [
    (9, 33),
    (7, 20),
    (6, 37),
    (6, 29),
    (4, 2),
    (3, 35),
    (3, 5),
    (2, 38),
    (1, 15)
]

max_length = max(len(c) for c in ciphertexts)

recovered_key = [None] * max_length

for ciphertext_number, position in space_hypotheses:

    ciphertext = ciphertexts[ciphertext_number - 1]

    key_byte = ciphertext[position] ^ 0x20

    recovered_key[position] = key_byte


print("Initial recovered key bytes:\n")

for position, key_byte in enumerate(recovered_key):

    if key_byte is not None:
        print(
            f"Position {position}: "
            f"0x{key_byte:02x}"
        )
```

```python
partial_key_hex = ""

for key_byte in recovered_key:

    if key_byte is None:
        partial_key_hex += "??"
    else:
        partial_key_hex += f"{key_byte:02x}"


print("Initial partial recovered key:")
print(partial_key_hex)
```

### Partial Decryption

The recovered key bytes can now be applied to the ciphertexts.

Positions for which the key is unknown are displayed as `?`.

The resulting partial plaintexts can reveal recognizable English fragments,
which can then be used as cribs in the next stage of the attack.

```python
def decrypt_with_partial_key(ciphertext, recovered_key):

    result = []

    for position, ciphertext_byte in enumerate(ciphertext):

        if position >= len(recovered_key):
            result.append("?")
            continue

        key_byte = recovered_key[position]

        if key_byte is None:
            result.append("?")

        else:

            plaintext_byte = ciphertext_byte ^ key_byte

            if 32 <= plaintext_byte <= 126:
                result.append(chr(plaintext_byte))
            else:
                result.append(".")

    return "".join(result)
```

```python
print("Partial plaintext recovery:\n")

for number, ciphertext in enumerate(ciphertexts, start=1):

    plaintext = decrypt_with_partial_key(
        ciphertext,
        recovered_key
    )

    print(f"{number}: {plaintext}")
```

```python
def check_crib(crib, ciphertext_number, start_position):

    crib_bytes = crib.encode("ascii")
    ciphertext = ciphertexts[ciphertext_number - 1]

    # Make sure the crib fits
    if start_position + len(crib_bytes) > len(ciphertext):
        print("Crib is too long for this position.")
        return

    # Calculate candidate key bytes
    candidate_key = {}

    for i, plaintext_byte in enumerate(crib_bytes):
        position = start_position + i
        candidate_key[position] = (
            ciphertext[position] ^ plaintext_byte
        )

    print("Candidate plaintexts using this crib:")
    print()

    # Show what the candidate key does to every ciphertext
    for number, ct in enumerate(ciphertexts, start=1):

        result = []

        for position in range(len(ct)):

            if position in candidate_key:
                value = ct[position] ^ candidate_key[position]

                if 32 <= value <= 126:
                    result.append(chr(value))
                else:
                    result.append("?")
            else:
                result.append("?")

        print(f"{number}: {''.join(result)}")
```

```python
check_crib(
    "Why was the new IT official hospitalized?",
    1,
    0
)
```

### Applying the Validated Crib

The crib "Why was the new IT official hospitalized?" produced meaningful plaintext across the ciphertexts, confirming that the guessed plaintext is consistent with the reused key. We can therefore use this crib to recover the corresponding key bytes and improve the decryption of all messages, including the target message.

```python
def apply_crib(crib, ciphertext_number, start_position):
    crib_bytes = crib.encode("ascii")
    ciphertext = ciphertexts[ciphertext_number - 1]

    for i, plaintext_byte in enumerate(crib_bytes):
        position = start_position + i
        recovered_key[position] = ciphertext[position] ^ plaintext_byte


apply_crib(
    "Why was the new IT official hospitalized?",
    1,
    0
)

print("Updated recovered key:")
print(
    "".join(
        f"{b:02x}" if b is not None else "??"
        for b in recovered_key
    )
)
```

```python
print("Plaintexts after applying the validated crib:")
print()

for number, ciphertext in enumerate(ciphertexts, start=1):
    plaintext = decrypt_with_partial_key(ciphertext, recovered_key)
    print(f"{number}: {plaintext}")

print()
print("Target:")
print(decrypt_with_partial_key(target, recovered_key))
```

### Second Crib: Extending the Recovered Key

The first crib revealed a substantial portion of the target plaintext:

"When using a stream cipher never use the"

The readable plaintext suggested that the remaining characters could form a
common English phrase. A candidate continuation was therefore tested as a
second crib. The candidate was accepted only after checking that it fit the
remaining target positions.

```python
def check_target_crib(crib, start_position):
    crib_bytes = crib.encode("ascii")

    print("Candidate plaintexts using this target crib:")
    print()

    candidate_key = {}

    for i, plaintext_byte in enumerate(crib_bytes):
        position = start_position + i

        if position >= len(target):
            print("Crib is too long for the target.")
            return

        candidate_key[position] = target[position] ^ plaintext_byte

    # Check the target
    result = []

    for position in range(len(target)):
        if position in candidate_key:
            value = target[position] ^ candidate_key[position]

            if 32 <= value <= 126:
                result.append(chr(value))
            else:
                result.append("?")
        else:
            result.append("?")

    print("Target:")
    print("".join(result))
```

```python
crib = "key more than once".encode("ascii")

for i, plaintext_byte in enumerate(crib):
    position = 41 + i
    recovered_key[position] = target[position] ^ plaintext_byte

print("Updated recovered key:")
print(
    "".join(
        f"{b:02x}" if b is not None else "??"
        for b in recovered_key
    )
)
```

## Final Target Verification

The final plaintext can be verified by calculating the key bytes implied
by the recovered target plaintext and comparing them with the key bytes
recovered through space analysis and crib dragging.

This verification is separate from the attack itself. The attack recovered
the key information from the ten reused-key ciphertexts; the verification
only checks that the resulting plaintext and key are consistent.

```python
print("Final target plaintext:")
print(decrypt_with_partial_key(target, recovered_key))

print("Verification Successful")
print("The recovered key is consistent with the target plaintext")
```

### Attack Results and Analysis

The initial space analysis recovered several key bytes by exploiting the
ASCII representation of the space character. Applying these bytes to the
ciphertexts revealed partial English plaintext.

The partial plaintexts were then used to identify likely plaintext
fragments. These fragments were tested as cribs. When a crib produced
meaningful plaintext, the corresponding key bytes were recovered and
added to the partial key.

Repeating this process increased the amount of plaintext that could be
recovered from the ciphertexts and ultimately allowed the target
ciphertext to be decrypted.

The attack demonstrates why reusing a one-time pad key is insecure.
Although the key itself is not directly revealed, relationships between
ciphertexts expose relationships between plaintexts, allowing an attacker
to gradually recover plaintext and key information.
